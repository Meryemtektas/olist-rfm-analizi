```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("olistbr/brazilian-ecommerce")

print("Path to dataset files:", path)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# kagglehub ile indirilen dosya yolunu alıyoruz
folder_path = path

# Klasörün içindeki tüm CSV dosyalarını görelim
print("İndirilen Dosyalar:")
print(os.listdir(folder_path))

# CSV dosyalarını pandas ile yüklüyoruz
orders = pd.read_csv(os.path.join(folder_path, 'olist_orders_dataset.csv'))
items = pd.read_csv(os.path.join(folder_path, 'olist_order_items_dataset.csv'))
customers = pd.read_csv(os.path.join(folder_path, 'olist_customers_dataset.csv'))
products = pd.read_csv(os.path.join(folder_path, 'olist_products_dataset.csv'))

# Veri setlerini birleştirip (Merge) ana veri tablosunu oluşturuyoruz
df = orders.merge(items, on='order_id', how='inner')
df = df.merge(customers, on='customer_id', how='inner')

print("\nVeri Başarıyla Birleştirildi! Satır ve Sütun Sayısı:", df.shape)

# İlk 5 satıra göz atalım
df.head()
```

```python
# 1. Tarih sütunlarını datetime formatına çevirelim
date_columns = ['order_purchase_timestamp', 'order_approved_at',
                'order_delivered_carrier_date', 'order_delivered_customer_date',
                'order_estimated_delivery_date']

for col in date_columns:
    df[col] = pd.to_datetime(df[col])

# 2. Toplam Harcama tutarını hesaplayalım (Ürün Fiyatı + Kargo Ücreti)
df['total_order_cost'] = df['price'] + df['freight_value']

# 3. Temel veri özetini görelim
print("--- Veri Bilgisi ve Veri Tipleri ---")
df.info()

print("\n--- Eksik Değer Sayıları ---")
print(df.isnull().sum())
```

```python
# 1. Sipariş Durumlarının Dağılımı (Teslim Edilen, İptal Edilen vs.)
plt.figure(figsize=(10, 4))
sns.countplot(data=df, x='order_status', palette='Set2')
plt.title('Sipariş Durumlarının Dağılımı')
plt.xlabel('Sipariş Durumu')
plt.ylabel('Sipariş Sayısı')
plt.show()

# 2. Aylık Ciro / Satış Trendi
# Sipariş tarihine göre ay bazında gruplama yapıyoruz
df['year_month'] = df['order_purchase_timestamp'].dt.to_period('M')
monthly_sales = df.groupby('year_month')['total_order_cost'].sum().reset_index()
monthly_sales['year_month'] = monthly_sales['year_month'].astype(str)

plt.figure(figsize=(12, 5))
sns.lineplot(data=monthly_sales, x='year_month', y='total_order_cost', marker='o', color='b')
plt.xticks(rotation=45)
plt.title('Aylara Göre Toplam Ciro Değişimi (Satış Trendi)')
plt.xlabel('Yıl - Ay')
plt.ylabel('Toplam Satış Tutarı (BRL)')
plt.grid(True)
plt.show()

# 3. Haftanın Günlerine Göre Sipariş Yoğunluğu
df['day_of_week'] = df['order_purchase_timestamp'].dt.day_name()
days_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

plt.figure(figsize=(9, 4))
sns.countplot(data=df, x='day_of_week', order=days_order, palette='viridis')
plt.title('Haftanın Günlerine Göre Sipariş Sayıları')
plt.xlabel('Günler')
plt.ylabel('Sipariş Sayısı')
plt.show()
```

```python
import datetime as dt

# 1. Referans tarihi belirleme (En son sipariş tarihinden 1 gün sonrası)
max_date = df['order_purchase_timestamp'].max() + dt.timedelta(days=1)

# 2. Müşteri bazında Recency, Frequency ve Monetary hesaplama
rfm = df.groupby('customer_unique_id').agg({
    'order_purchase_timestamp': lambda x: (max_date - x.max()).days, # Recency
    'order_id': 'nunique',                                           # Frequency
    'total_order_cost': 'sum'                                        # Monetary
}).reset_index()

# Sütun isimlerini düzenleyelim
rfm.columns = ['customer_unique_id', 'Recency', 'Frequency', 'Monetary']

# 3. RFM Skorlarını Oluşturma (1-5 arası puanlama)
# Recency için gün sayısı AZ olan daha yüksek puan alır (ascending=False)
rfm['R_Score'] = pd.qcut(rfm['Recency'], q=5, labels=[5, 4, 3, 2, 1])

# Frequency'de çoğu müşteri 1 sipariş verdiği için rank metodunu kullanıyoruz
rfm['F_Score'] = pd.qcut(rfm['Frequency'].rank(method='first'), q=5, labels=[1, 2, 3, 4, 5])

# Monetary için harcama tutarına göre puanlama
rfm['M_Score'] = pd.qcut(rfm['Monetary'], q=5, labels=[1, 2, 3, 4, 5])

# RFM Skorunu birleştirme
rfm['RFM_Score'] = rfm['R_Score'].astype(str) + rfm['F_Score'].astype(str) + rfm['M_Score'].astype(str)

# 4. Müşterileri Segmentlere Ayırma Mantığı
def segment_customer(df):
    r = int(df['R_Score'])
    f = int(df['F_Score'])

    if r >= 4 and f >= 4:
        return 'Champions (Şampiyonlar)'
    elif r >= 3 and f >= 3:
        return 'Loyal Customers (Sadık Müşteriler)'
    elif r >= 3 and f <= 2:
        return 'Promising / New (Gelecek Vaat Eden / Yeni)'
    elif r <= 2 and f >= 3:
        return 'At Risk (Riske Girmiş)'
    else:
        return 'Hibernating / Lost (Kayıp Müşteriler)'

rfm['Segment'] = rfm.apply(segment_customer, axis=1)

# İnceleyelim
rfm.head()
```

```python
# Müşteri Segmentlerinin Dağılımını Grafikleştirme
plt.figure(figsize=(10, 5))
segment_counts = rfm['Segment'].value_counts()

sns.barplot(x=segment_counts.values, y=segment_counts.index, palette='magma')
plt.title('Müşteri Segmentlerinin Dağılımı (RFM Analizi)')
plt.xlabel('Müşteri Sayısı')
plt.ylabel('Segmentler')
plt.show()

# Özet Tablo
print("--- Segment Başına Müşteri Sayıları ve Ortalamalar ---")
rfm.groupby('Segment').agg({
    'Recency': 'mean',
    'Frequency': 'mean',
    'Monetary': ['mean', 'count']
}).round(2)
```
